# 03 Regime Detection

Fit the Hidden Markov Model as the primary regime detector and compare it with a GMM baseline.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

SRC_PATH = PROJECT_ROOT / "src"
if str(SRC_PATH) not in sys.path:
    sys.path.append(str(SRC_PATH))

import pandas as pd
from regime_detection import GaussianMixtureRegimeDetector, HiddenMarkovRegimeDetector
from utils import project_path
```

```python
feature_frame = pd.read_csv(project_path("data", "processed", "SPY_features.csv"), index_col=0, parse_dates=True)

hmm_detector = HiddenMarkovRegimeDetector(n_regimes=4, random_state=42)
hmm_regimes = hmm_detector.fit_predict(feature_frame)
hmm_regimes.to_csv(project_path("data", "processed", "SPY_regimes.csv"))
hmm_detector.transition_matrix().to_csv(project_path("data", "processed", "SPY_hmm_transition_matrix.csv"))

pd.DataFrame([summary.__dict__ for summary in hmm_detector.summarize_regimes()])
```

```python
gmm_detector = GaussianMixtureRegimeDetector(n_regimes=4, random_state=42)
gmm_regimes = gmm_detector.fit_predict(feature_frame)
gmm_regimes.to_csv(project_path("data", "processed", "SPY_regimes_gmm.csv"))

pd.DataFrame([summary.__dict__ for summary in gmm_detector.summarize_regimes()])
```
