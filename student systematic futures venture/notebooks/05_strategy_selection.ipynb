# 05 Strategy Selection

This notebook maps directional probabilities and market context into specific derivatives structures. The rules are deliberately transparent so they can be reviewed, challenged, and adapted for small-capital trading constraints.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.options_strategy import OptionsStrategySelector
from src.utils import load_yaml_config

config = load_yaml_config(PROJECT_ROOT / "config" / "parameters.yaml")
processed_dir = PROJECT_ROOT / config["data"]["processed_data_dir"]
signals = pd.read_csv(processed_dir / "nsei_signals.csv", index_col=0, parse_dates=True)

selector = OptionsStrategySelector(**config["options_strategy"])
strategy_frame = selector.generate_strategy_signals(signals)
strategy_frame.to_csv(processed_dir / "nsei_strategy_signals.csv")
```
