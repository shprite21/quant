# 01 Data Exploration

Use this notebook to inspect the cleaned price history, return distributions, and benchmark behavior before strategy research.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

project_root = Path('..').resolve()
prices = pd.read_csv(project_root / 'data' / 'processed' / 'clean_prices.csv', index_col=0, parse_dates=True)
returns = pd.read_csv(project_root / 'data' / 'processed' / 'daily_returns.csv', index_col=0, parse_dates=True)
prices.head()
```

```python
(prices / prices.iloc[0]).plot(figsize=(12, 6), title='Normalized Price Paths')
plt.show()
```
