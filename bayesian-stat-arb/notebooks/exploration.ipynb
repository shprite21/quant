```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from src.data_loader import MarketDataLoader
from src.cointegration import CointegrationTester
from src.utils import load_config

config = load_config(PROJECT_ROOT / 'configs' / 'strategy_config.yaml')
loader = MarketDataLoader.from_config(config, PROJECT_ROOT)
prices = loader.load_or_download(use_cached=True)
prices.tail()
```
